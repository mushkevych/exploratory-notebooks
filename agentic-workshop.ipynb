## OpenClaw

1. Run `docker compose -f docker-compose-openclaw.yml up -d` to start OpenClaw.
2. Open http://localhost:3030 in the browser with "change-this-long-random-token" as the token.

local files
```text
Create a local Python script at /home/node/.openclaw/workspace/analyze_numbers.py.

The script should:
1. Create a list of numbers: [10, 20, 30, 40, 50]
2. Calculate the count, sum, mean, min, and max
3. Write the results to /home/node/.openclaw/workspace/results.txt

After creating the script, run it with python3, then read results.txt and summarize the output.
```

```text
Create a local CSV file at `$HOME/workspace/contacts.csv`.

The file should contain 20 fictional contacts with these columns:
- name
- phone
- email
- company
- note

Use obviously fake phone numbers and emails, for example emails under example.com.
Do not truncate or compact - write full file.

Then:
1. Read the CSV file back.
2. Find the contact named "Maya Chen".
3. Create a draft email to Maya Chen at `$HOME/workspace/draft_email_to_maya.txt`.
4. The email should invite her to a short OpenClaw demo next week.
5. Read the draft email back and summarize what you created.

Do not actually send any email.
Only create the local files.
```
